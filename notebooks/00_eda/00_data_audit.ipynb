```python
import duckdb
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "data" / "raw").exists():   # notebook runs from notebooks/00_eda/
    ROOT = ROOT.parents[1]                 # step up to the repo root

for f in sorted((ROOT / "data" / "raw").glob("*.csv")):
    print(f"\n=== {f.name} ===")
    print(duckdb.sql(f"DESCRIBE SELECT * FROM read_csv_auto('{f.as_posix()}')"))
```

```python
hh = (ROOT / "data" / "raw" / "hh_demographic.csv").as_posix()
for c in ["classification_1", "classification_2", "classification_3",
          "classification_4", "classification_5"]:
    vals = duckdb.sql(f"SELECT DISTINCT {c} FROM read_csv_auto('{hh}') ORDER BY 1").df()
    print(c, "->", vals[c].tolist())
```

```python
for f in sorted((ROOT / "data" / "raw").glob("*.csv")):
    n = duckdb.sql(f"SELECT COUNT(*) FROM read_csv_auto('{f.as_posix()}')").fetchone()[0]
    print(f"{f.name:25s} {n:>12,}")
```
